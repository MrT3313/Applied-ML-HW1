# Imports

```python
import joblib
import pandas as pd
from sklearn.metrics import mean_squared_error

import Q1.consts as Q1_CONSTS

from utils import read_csv_to_df

pd.set_option('display.float_format', '{:.4f}'.format)
```

# Fetch Data

```python
test_df = read_csv_to_df(path=Q1_CONSTS.TEST_PATH)

ATTRIBUTES = [column for column in test_df.columns if column != Q1_CONSTS.TARGET]
X_test, y_test = test_df[ATTRIBUTES], test_df[Q1_CONSTS.TARGET]
```

# Method

- **Models:** the two models D and E selected and saved, loaded from `Q1/MODELS` as they are: D's linear regression (`linear_regression.joblib`) and E's polynomial regression (`polynomial_regression.joblib`). Each is one pipeline that takes the 8 raw attributes and builds its own columns, so nothing is fit on the test rows.
- **Metric:** RMSE in `Life Ladder` points on every test row, the metric D and E used.
- **Test data used once:** each model's RMSE is reported as it comes. Neither score is used to pick, change or refit a model.

# Load the Models

```python
MODELS = {  # the models D and E selected and saved, as they are
	'D: linear regression': joblib.load(Q1_CONSTS.MODEL_OUTPUT_DIR / 'linear_regression.joblib'),
	'E: polynomial regression': joblib.load(Q1_CONSTS.MODEL_OUTPUT_DIR / 'polynomial_regression.joblib'),
}
```

# Test RMSE

```python
PREDICTIONS = {name: model.predict(X_test) for name, model in MODELS.items()}

pd.DataFrame({'test rmse': {name: mean_squared_error(y_test, predicted, squared=False) for name, predicted in PREDICTIONS.items()}})
```

### Notes / Observations

- Both saved models predict every test row as D and E saved them, with nothing refit.
- E's polynomial model misses by less than D's linear model on new rows: 0.5024 against 0.5251 `Life Ladder` points.
